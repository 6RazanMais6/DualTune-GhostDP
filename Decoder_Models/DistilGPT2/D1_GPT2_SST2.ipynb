```
from pathlib import Path
script_path = Path('/content/D1_GPT2_SST2_multiseed.py')
script_path.write_text('#!/usr/bin/env python3\n"""GPT-2 SST-2 reproducible experiment. Run under the Python 3.12 Colab venv."""\nimport argparse, os, time, random, json, csv, gc\nfrom pathlib import Path\nimport numpy as np\nimport torch\nimport torch.nn as nn\nimport torch.nn.functional as F\nfrom torch.utils.data import DataLoader\nfrom transformers import GPT2TokenizerFast, GPT2LMHeadModel\nfrom datasets import load_dataset, Dataset\nfrom private_transformers import PrivacyEngine\nfrom tqdm.auto import tqdm\nimport psutil\nfrom datetime import datetime\n\nLOG_FILE = None\n\ndef log(message):\n    stamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")\n    line = f"[{stamp}] {message}"\n    print(line, flush=True)\n    if LOG_FILE is not None:\n        with open(LOG_FILE, "a", encoding="utf-8") as f:\n            f.write(line + "\\n")\n\ndef progress(seed, epoch, step, total, loss, started, train_acc):\n    done = step + 1\n    elapsed = time.monotonic() - started\n    eta = elapsed / done * (total - done)\n    mem = torch.cuda.memory_allocated() / 2**20\n    log(f"Seed {seed} | DP epoch {epoch}/{DP_EPOCHS} | "\n        f"{done}/{total} batches ({done/total:.1%}) | "\n        f"loss={loss:.4f} | TrainAcc={train_acc:.4f} | elapsed={elapsed/60:.1f} min | "\n        f"ETA={eta/60:.1f} min | GPU allocated={mem:.0f} MB")\n\n\nMODEL_NAME=\'gpt2\'; MAX_LEN=128; BATCH_SIZE=24\nPHASE1_EPOCHS=2; DP_EPOCHS=2; TARGET_EPSILON=3.0\nMAX_GRAD_NORM=0.5; LR_P1=2e-4; LR_DP=2e-4\nPOS_STR=\' positive\'; NEG_STR=\' negative\'\n\ndef seed_all(seed):\n    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)\n    torch.cuda.manual_seed_all(seed)\n    torch.backends.cudnn.deterministic=True\n    torch.backends.cudnn.benchmark=False\n\ndef balance(ds,seed):\n    # Exact Python random.sample sequence from original uploaded notebook.\n    a=[ex for ex in ds if ex[\'label\']==0]\n    b=[ex for ex in ds if ex[\'label\']==1]\n    random.seed(seed)\n    n=min(len(a),len(b))\n    chosen=random.sample(a,n)+random.sample(b,n)\n    random.shuffle(chosen)\n    return Dataset.from_list(chosen)\n\ndef collate(batch):\n    return {k:torch.tensor([x[k] for x in batch],dtype=torch.long)\n            for k in (\'input_ids\',\'attention_mask\',\'label_token\')}\n\ndef last_logits(model,ids,mask):\n    logits=model(input_ids=ids,attention_mask=mask).logits\n    return logits[torch.arange(ids.shape[0],device=ids.device),\n                  (mask.sum(1)-1).clamp(min=0),:]\n\ndef eval_acc(model,loader,device,pos,neg):\n    model.eval(); good=total=0\n    with torch.no_grad():\n        for batch in loader:\n            ids=batch[\'input_ids\'].to(device); mask=batch[\'attention_mask\'].to(device)\n            target=batch[\'label_token\'].to(device)\n            logits=last_logits(model,ids,mask)\n            pred=torch.where(logits[:,pos]>=logits[:,neg],pos,neg)\n            good+=(pred==target).sum().item(); total+=len(target)\n    model.train()\n    return good/total\n\ndef setup_data(seed):\n    raw=load_dataset(\'nyu-mll/glue\',\'sst2\')\n    train=balance(raw[\'train\'],seed); val=balance(raw[\'validation\'],seed)\n    tok=GPT2TokenizerFast.from_pretrained(MODEL_NAME)\n    tok.pad_token=tok.eos_token\n    pos=tok.encode(POS_STR,add_special_tokens=False)\n    neg=tok.encode(NEG_STR,add_special_tokens=False)\n    assert len(pos)==len(neg)==1\n    pos,neg=pos[0],neg[0]\n    def encode(ex):\n        encoded=tok(f"Review: {ex[\'sentence\']}\\nSentiment:",padding=\'max_length\',\n                    truncation=True,max_length=MAX_LEN)\n        return {\'input_ids\':encoded[\'input_ids\'],\n                \'attention_mask\':encoded[\'attention_mask\'],\n                \'label_token\':pos if ex[\'label\']==1 else neg}\n    val_tok=val.map(encode)\n    val_loader=DataLoader(val_tok,batch_size=BATCH_SIZE,shuffle=False,collate_fn=collate)\n    return train,encode,val_loader,tok,pos,neg\n\ndef build_model(tok,weight_tying,device):\n    m=GPT2LMHeadModel.from_pretrained(MODEL_NAME)\n    m.config.pad_token_id=tok.pad_token_id\n    if weight_tying:\n        m.tie_weights()\n    else:\n        m.lm_head.weight=nn.Parameter(m.lm_head.weight.detach().clone())\n    m.to(device)\n    tied=(m.lm_head.weight.data_ptr()==m.transformer.wte.weight.data_ptr())\n    assert tied==weight_tying,f\'WT check failed: {tied}\'\n    return m\n\ndef resource_start():\n    torch.cuda.empty_cache(); torch.cuda.reset_peak_memory_stats()\n    return time.time(),psutil.Process().memory_info().rss/2**20\n\ndef resource_end(start):\n    t,rss=start\n    return dict(time_min=(time.time()-t)/60,\n                gpu_peak_allocated_mb=torch.cuda.max_memory_allocated()/2**20,\n                gpu_peak_reserved_mb=torch.cuda.max_memory_reserved()/2**20,\n                cpu_rss_delta_mb=psutil.Process().memory_info().rss/2**20-rss)\n\ndef train_dp(model,train_tok,val_loader,device,pos,neg,clipping,seed):\n    loader=DataLoader(train_tok,batch_size=BATCH_SIZE,shuffle=True,drop_last=True,collate_fn=collate)\n    opt=torch.optim.Adam(model.parameters(),lr=LR_DP)\n    pe=PrivacyEngine(model,batch_size=BATCH_SIZE,sample_size=len(train_tok),\n                     epochs=DP_EPOCHS,max_grad_norm=MAX_GRAD_NORM,\n                     target_epsilon=TARGET_EPSILON,clipping_mode=clipping)\n    pe.attach(opt)\n    log(f"Seed {seed} | PrivacyEngine attached | clipping={clipping} | {len(loader)} batches/epoch")\n    results=[]\n    for epoch in range(DP_EPOCHS):\n        model.train(); total_loss=0; train_correct=0; train_seen=0\n        epoch_start=time.monotonic()\n        log(f"Seed {seed} | START DP epoch {epoch+1}/{DP_EPOCHS}")\n        for step,batch in enumerate(loader):\n            ids=batch[\'input_ids\'].to(device); mask=batch[\'attention_mask\'].to(device)\n            target=batch[\'label_token\'].to(device)\n            opt.zero_grad()\n            logits=last_logits(model,ids,mask)\n            loss=F.cross_entropy(logits,target,reduction=\'none\')\n            pred=torch.where(logits[:,pos]>=logits[:,neg],pos,neg)\n            train_correct+=(pred.detach()==target).sum().item()\n            train_seen+=target.numel()\n            opt.step(loss=loss)\n            total_loss+=loss.detach().mean().item()\n            if step==0 or (step+1)%100==0 or step+1==len(loader):\n                progress(seed,epoch+1,step,len(loader),loss.mean().item(),epoch_start,train_correct/train_seen)\n        acc=eval_acc(model,val_loader,device,pos,neg)\n        log(f\'Seed {seed} | END DP epoch {epoch+1} | mean_loss={total_loss/len(loader):.4f} | TrainAcc={train_correct/train_seen:.4f} | val_acc={acc:.4f} | elapsed={(time.monotonic()-epoch_start)/60:.1f} min\')\n        results.append({\'epoch\':epoch+1,\'loss\':total_loss/len(loader),\'validation_accuracy\':acc})\n    # Do not pass epoch count as `steps`; use actual optimizer steps for privacy accounting.\n    spent=None\n    try:\n        spent=pe.get_privacy_spent(steps=DP_EPOCHS*len(loader))\n    except Exception as e:\n        log(f\'Seed {seed} | Privacy accountant query failed: {e!r}\')\n    return results,repr(spent)\n\ndef save_result(args,config,seed,model,metrics,epochs,privacy):\n    out=Path(args.output_dir)/config/f\'seed_{seed}\';out.mkdir(parents=True,exist_ok=True)\n    torch.save(model.state_dict(),out/\'final_model.pth\')\n    record={\'configuration\':config,\'seed\':seed,\'gpu\':torch.cuda.get_device_name(0),\n            \'target_epsilon\':TARGET_EPSILON,\'max_grad_norm\':MAX_GRAD_NORM,\n            \'batch_size\':BATCH_SIZE,\'dp_epochs\':DP_EPOCHS,\'privacy_report\':privacy,\n            \'validation_accuracy\':epochs[-1][\'validation_accuracy\'],**metrics,\n            \'epoch_metrics\':epochs}\n    (out/\'results.json\').write_text(json.dumps(record,indent=2))\n    with (Path(args.output_dir)/f\'{config}_summary.csv\').open(\'a\',newline=\'\') as f:\n        row={k:v for k,v in record.items() if k!=\'epoch_metrics\'}\n        writer=csv.DictWriter(f,fieldnames=list(row))\n        if f.tell()==0:writer.writeheader()\n        writer.writerow(row)\n    log(f\'Seed {seed} | SAVED {out}\')\n\ndef parse():\n    p=argparse.ArgumentParser()\n    p.add_argument(\'--seeds\',nargs=\'+\',type=int,default=[60])\n    p.add_argument(\'--output-dir\',default=\'/content/drive/MyDrive/DualTune_GhostDP/multiseed\')\n    return p.parse_args()\n\n\ndef main():\n    args=parse();Path(args.output_dir).mkdir(parents=True,exist_ok=True)\n    assert torch.cuda.is_available(),\'Colab GPU required\'\n    device=torch.device(\'cuda\')\n    global LOG_FILE\n    LOG_FILE=Path(args.output_dir)/\'D1_progress.log\'\n    log(f\'CONFIG D1 | WT=True | clipping=default | GPU={torch.cuda.get_device_name(0)} | seeds={args.seeds}\')\n    for seed in args.seeds:\n        log(f\'===== START SEED {seed} =====\')\n        seed_start=time.monotonic()\n        seed_all(seed)\n        log(f\'Seed {seed} | Loading, balancing, and tokenizing SST-2\')\n        train,encode,val_loader,tok,pos,neg=setup_data(seed)\n        log(f\'Seed {seed} | Dataset ready: {len(train)} training examples\')\n        out=Path(args.output_dir)/\'D1\'/f\'seed_{seed}\'\n        out.mkdir(parents=True,exist_ok=True)\n        log(f\'Seed {seed} | Loading GPT-2 (WT=True)\')\n        model=build_model(tok,True,device)\n        log(f\'Seed {seed} | Tokenizing training examples\')\n        train_tok=train.map(encode)\n        log(f\'Seed {seed} | Training dataset tokenized\')\n        start=resource_start()\n        epochs,privacy=train_dp(model,train_tok,val_loader,device,pos,neg,\'default\',seed)\n        metrics=resource_end(start)\n        log(f\'Seed {seed} | Saving final model and results\')\n        save_result(args,\'D1\',seed,model,metrics,epochs,privacy)\n        log(f\'Seed {seed} | COMPLETE | total elapsed={(time.monotonic()-seed_start)/60:.1f} min\')\n        del model,train_tok,train,val_loader,tok;gc.collect();torch.cuda.empty_cache()\n\nif __name__==\'__main__\':main()\n')
print('Wrote:', script_path, '| bytes:', script_path.stat().st_size)
```
